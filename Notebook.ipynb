```python
from langchain.document_loaders import PyPDFLoader
# === 3. Charger les fichiers PDF ===
pdf_paths = ["/content/drive/MyDrive/Group9_NLP/Review-Year-2023-2024.pdf"]  # remplace par tes chemins locaux
pdf_docs = []
for path in pdf_paths:
    loader = PyPDFLoader(path)
    pdf_docs.extend(loader.load())
```

```python
# === 4. Fusionner les documents ===
all_docs =pdf_docs
all_docs[40]
```

```python

from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=1000)
docs = text_splitter.split_documents(all_docs)
```

```python
from langchain.embeddings import HuggingFaceEmbeddings
embeddings= HuggingFaceEmbeddings()

vector= embeddings.embed_query("hello,word!")
```

```python
from langchain_chroma import Chroma
vectorstore=Chroma.from_documents(docs,embeddings)
```

```python
retriever= vectorstore.as_retriever(search_type="similarity",search_kwargs={"k":2})
retriever_docs=retriever.invoke("academic director")
```

```python
from langchain.llms import HuggingFaceHub
from langchain_huggingface import HuggingFacePipeline
from langchain.prompts import PromptTemplate
from transformers import pipeline
from langchain_core.output_parsers import StrOutputParser
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
model_id="tiiuae/falcon-7b-instruct"
```

```python
text_generation_pipeline=pipeline(
    "text-generation",
    model=model_id,
    model_kwargs={"torch_dtype":torch.bfloat16},max_new_tokens=400
)
llm=HuggingFacePipeline(pipeline=text_generation_pipeline)

```

```python

from langchain_core.runnables import RunnablePassthrough
# === PROMPT TEMPLATE ===
prompt_template = """
<|system|>
Answer the question based on your knowledge. Use the following context to help :

{context}
</s>
<|user|>
{question}

<|assistant|>
"""
prompt = PromptTemplate(template=prompt_template, input_variables=["context", "question"])

# === CHAÎNE LLM ===
llm_chain = prompt | llm | StrOutputParser()

# === CHAÎNE RAG ===
rag_chain = {"context": retriever, "question": RunnablePassthrough()} | llm_chain
```

```python
from langchain.evaluation import load_evaluator
from langchain.evaluation import EvaluatorType
```

```python

# === QUESTION À POSER ===
question = "What is the program of master?"

# === GÉNÉRATION DU RÉSULTAT PAR LE LLM ===
prediction = rag_chain.invoke(question)

# === RÉFÉRENCE ATTENDUE ===
reference = (
   "The AIMS Master's in Mathematical Sciences is a structured program offering a Master’s II degree designed for students with at least a Master’s I or four years of undergraduate studies. Applicants must have a background in  mathematics, physics, computer science, or related fields. The program is reviewed annually by the academic council and may be adjusted based on feedback from academic and industry stakeholders. AIMS Senegal offers a Master’s in Mathematical Sciences with three options: 1. A Classical or Regular Program oriented toward research, 2. A Cooperative Program focused on industry, including an internship phase, 3. A Machine Learning & Artificial Intelligence Program (AMMI). Courses are organized into teaching units, and students are trained to apply mathematical problem-solving methods to various key areas of development. At the end of the academic year, students are required to present an oral defense of their research project, regardless of their are specialization, followed by a graduation ceremony."
)

# === CRITÈRES DE SCORE (exemple simple) ===
score_criteria = {"helpfulness": "How helpful is the answer in addressing the user's question?"}

# === ÉVALUATEUR ===
evaluator = load_evaluator(
    "labeled_score_string",
    criteria=score_criteria,
    llm=llm
)

# === ÉVALUATION ===
eval_result = evaluator.evaluate_strings(
    input=question,
    prediction=prediction,
    reference=reference
)

# === AFFICHAGE DU RÉSULTAT ===
print("LLM's answer:")
print(prediction)
print("\nEvaluation reasoning:")
print(eval_result["reasoning"])
print("\nScore(s):")
print(eval_result["score"])
```
